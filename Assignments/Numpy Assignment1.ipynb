```python
#c. Append a new mobile having price of 13999 Rs. to this array
arr = np.append(arr,13999)
arr
```

```python
 #d. Reverse this array of mobile phones prices
arr[::-1]
```

```python
#f. Sort the array in descending order of price
arr = np.sort(arr)[::-1]
arr
```
